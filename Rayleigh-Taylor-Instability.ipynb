$$\begin{align*}
&\textbf{Objective:} \\
&\\
&\text{The goal is to create a mathematical model, simulating the Rayleigh-Taylor} \\
&\text{instability between two noncompressible fluids with different densities.} \\
&\\
&\text{When observing from an undisturbed state which is essentially equilibrium where one fluid} \\
&\text{perfectly lies on top of the other. If there is even a small displacement, you would see} \\
&\text{the heavier fluid sink, while the lighter fluid rises. This growth further develops until we} \\
&\text{get the formation of rising columns of light fluid known as bubbles, and falling columns of} \\
&\text{the heavier fluid known as spikes. At some point during the perturbation, growth becomes} \\
&\text{unstable and my goal is to find a rough estimate as to where that point is across} \\
&\text{different parameters.} \\
&\\
&\text{For the reference paper, I will be using an article written by Jason Oakley} \\
&\text{from the year 2004. In this article he derived an analytical solution to the} \\
&\text{problem that I am trying to solve. So using that, my simulation will track the} \\
&\text{growth of an initial interface perturbation across a range of Atwood numbers} \\
&\text{and wavenumbers.}
\end{align*}$$

$$\begin{align*}
&\textbf{Assumptions:} \\
&\\
&1. \quad \text{2-dimensional, i.e., } \frac{\partial}{\partial y} = 0 \\
&2. \quad \text{incompressibility for a given layer of fluid} \\
&3. \quad \text{velocity rewritten in terms of initial constant values plus small deviations} \\
&4. \quad \text{inviscid} \\
&5. \quad \text{initial velocities are zero} \\
&6. \quad \text{momentum equation is linearized} \\
&7. \quad \text{neglect energy equation (isothermal, etc.)}
\end{align*}$$

$$\begin{align*}
&\textbf{Mathematical Formulation:} \\
&\\
&1. \quad \text{Conservation of Mass (Continuity):} \\
&\quad \frac{\partial \rho}{\partial t} + \nabla \cdot (\rho \mathbf{V}) = 0 \\
&\\
&2. \quad \text{Conservation of Momentum:} \\
&\quad \rho \frac{D\mathbf{V}}{Dt} = \rho \mathbf{g} - \nabla p + \nabla \cdot \boldsymbol{\tau} \\
&\\
&\text{where the material derivative is defined as:} \\
&\quad \frac{D}{Dt} = \frac{\partial}{\partial t} + U \frac{\partial}{\partial x} + V \frac{\partial}{\partial y} + W \frac{\partial}{\partial z} \\
&\\
&\text{After applying the assumptions above, Oakley derives the following system of four equations:} \\
&\\
&1. \quad \frac{\partial u}{\partial x} + \frac{\partial w}{\partial z} = 0 \\
&\\
&2. \quad \rho \frac{\partial u}{\partial t} = -\frac{\partial \delta p}{\partial x} \\
&\\
&3. \quad \rho \frac{\partial w}{\partial t} = -\delta \rho g - \frac{\partial \delta p}{\partial z} \\
&\\
&4. \quad \frac{\partial \delta \rho}{\partial t} = -w \frac{\partial \rho}{\partial z} \\
&\\
&\text{with unknowns } u, w, \delta p, \text{ and } \delta \rho
\end{align*}$$

$$\begin{align*}
&\textbf{Variables:} \\
&\\
&1. \quad u, w \quad \text{Velocity perturbations in } x \text{ and } z \\
&\\
&2. \quad \delta p \quad \text{Pressure perturbation} \\
&\\
&3. \quad \delta \rho \quad \text{Density perturbation} \\
&\\
&4. \quad \rho_1, \rho_2 \quad \text{Density of lower and upper fluid} \\
&\\
&5. \quad \eta(x,t) \quad \text{Interface displacement} \\
&\\
&6. \quad \eta_0 \quad \text{Initial perturbation amplitude} \\
&\\
&7. \quad n \quad \text{Growth rate (eigenvalue)} \\
&\\
&8. \quad k \quad \text{Wavenumber} \\
&\\
&9. \quad g \quad \text{Gravitational acceleration}
\end{align*}$$

$$\begin{align*}
&\textbf{Parameters:} \\
&\\
&1. \quad A \quad \text{Atwood number} \\
&\\
&2. \quad k \quad \text{Wavenumber } (= 2\pi/\lambda) \\
&\\
&3. \quad \eta_0 \quad \text{Initial amplitude} \\
&\\
&4. \quad L_x, L_z \quad \text{Domain dimensions} \\
&\\
&5. \quad g \quad \text{Gravity}
\end{align*}$$

$$\begin{align*}
&\textbf{Linearizing the System} \\
&\\
&\text{Normal mode form:} \\
&q(x,z,t) = \hat{q}(z) \exp(ikx + nt) \\
&\\
&\text{Substituting into the four-equation reduced system converts the PDEs into ODEs in } z\text{:} \\
&\\
&\text{(1)} \quad iku + \frac{dw}{dz} = 0 \\
&\text{(2)} \quad \rho n u = -ik\delta p \\
&\text{(3)} \quad \rho n w = -\delta \rho g - \frac{d\delta p}{dz} \\
&\text{(4)} \quad n\delta \rho = -w\frac{d\rho}{dz} \\
&\\
&\text{Combining equations (1)-(4) yields the governing differential equation:} \\
&\frac{d}{dz}\left(\rho\frac{dw}{dz}\right) - k^2\rho w = \frac{k^2 g}{n^2}\frac{d\rho}{dz}w
\end{align*}$$

$$\begin{align*}
&\textbf{Linear Stability Analysis} \\
&\\
&\text{The Jacobian matrix:} \\
&\mathbf{J} = \begin{pmatrix} 0 & 1 \\ gkA & 0 \end{pmatrix} \\
&\\
&\text{Fixed point: } (0,0) \text{ (equilibrium - Undisturbed state)} \\
&\\
&\text{The eigenvalues are:} \\
&\lambda = \pm\sqrt{gkA} \\
&\\
&\text{For parameters } g = 9.81, \, k = 6.28, \, A = 0.20\text{:} \\
&\lambda_1 = +3.51 \quad \text{(unstable eigenvalue)} \\
&\lambda_2 = -3.51 \quad \text{(stable eigenvalue)} \\
&\\
&\textbf{Physical Interpretation:} \\
&\bullet \quad \text{For } A > 0 \text{ (heavy over light)} \to \text{real positive eigenvalue} \to \text{saddle point} \to \textbf{unstable} \text{ (RT instability)} \\
&\bullet \quad \text{For } A < 0 \text{ (light over heavy)} \to \text{purely imaginary eigenvalues} \to \text{centre point} \to \textbf{stable oscillations} \\
&\bullet \quad \text{For } A = 0 \text{ (equal densities)} \to \text{zero eigenvalues} \to \textbf{neutrally stable}
\end{align*}$$

$$\begin{align*}
&\textbf{Numerical Simulations/Data Analysis:}
\end{align*}$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from scipy.optimize import curve_fit
```

```python
#parameters
g = 9.81
rho_1 = 1.0
rho_2 = 1.5
A = (rho_2 - rho_1) / (rho_2 + rho_1)
lam = 1.0
eta_0 = 0.05 * lam
k = 2 * np.pi / lam
Lx = 1 * lam
Lz = 4 * lam
t_end = 0.6
```

```python
k_range = np.linspace(0, 4 * k, 500)
n = np.sqrt(g * k_range * A)

plt.figure(figsize=(8, 5))
plt.plot(k_range, n)
plt.xlabel('Wavenumber k')
plt.ylabel('Growth rate n')
plt.title(f'Dispersion Relation (A = {A:.2f})')
plt.axvline(x=k, color='r', linestyle='--', label=f'k = {k:.2f}')
plt.axhline(y=np.sqrt(g * k * A), color='g', linestyle='--', 
            label=f'n = {np.sqrt(g * k * A):.2f}')
plt.legend()
plt.grid(True)
plt.show()
```

```python
n_theory = np.sqrt(g * k * A)
t = np.linspace(0, 1.5, 500)
eta = eta_0 * np.exp(n_theory * t)
linear_limit = 0.1 * lam

plt.figure(figsize=(8, 5))
plt.plot(t, eta, label=r'$\eta(t) = \eta_0 e^{nt}$')
plt.axhline(y=linear_limit, color='r', linestyle='--', 
            label='Linear regime limit (0.1λ)')
plt.xlabel('Time t (s)')
plt.ylabel('Interface amplitude η(t)')
plt.title(f'Theoretical Interface Growth (A = {A:.2f}, n = {n_theory:.2f})')
plt.legend()
plt.grid(True)
plt.show()
```

```python
x = np.linspace(0, lam, 500)
eta_x = eta_0 * np.cos(k * x)

plt.figure(figsize=(8, 4))
plt.plot(x, eta_x, 'b-', label=r'$\eta(x) = \eta_0\cos(kx)$')
plt.axhline(y=0, color='k', linestyle='-', linewidth=0.8, 
            label='Unperturbed interface')
plt.axhline(y=eta_0, color='r', linestyle='--', linewidth=0.8, 
            label=f'η₀ = {eta_0:.3f}')
plt.axhline(y=-eta_0, color='r', linestyle='--', linewidth=0.8)
plt.fill_between(x, eta_x, Lz/2, alpha=0.15, color='blue', 
                 label=f'ρ₂ = {rho_2} (upper fluid)')
plt.fill_between(x, -Lz/2, eta_x, alpha=0.15, color='red', 
                 label=f'ρ₁ = {rho_1} (lower fluid)')
plt.xlabel('x')
plt.ylabel('η(x)')
plt.title('Initial Interface Perturbation at t = 0')
plt.ylim(-0.1, 0.1)
plt.legend(loc='upper right')
plt.grid(True)
plt.show()
```

```python
def linear_rt(t, eta_vec):
    eta, eta_dot = eta_vec
    return [eta_dot, (g * k * A) * eta]

eta_vec_0 = [eta_0, 0.0]

sol = solve_ivp(linear_rt, [0, t_end], eta_vec_0,
                max_step=0.001, dense_output=True)

t_sol = sol.t
eta_sol = sol.y[0]
eta_theory_sol = eta_0 * np.exp(n_theory * t_sol)

plt.figure(figsize=(8, 5))
plt.plot(t_sol, eta_theory_sol, 'r--', linewidth=2, 
         label=r'Theory: $\eta_0 e^{nt}$')
plt.plot(t_sol, eta_sol, 'b-', linewidth=2, 
         label='Numerical ODE solution')
plt.axhline(y=0.1 * lam, color='g', linestyle='--', 
            label='Linear regime limit (0.1λ)')
plt.xlabel('Time t (s)')
plt.ylabel('Interface Amplitude η(t)')
plt.title(f'Linear Stability Validation (A = {A:.2f}, n = {n_theory:.2f})')
plt.legend()
plt.grid(True)
plt.show()

relative_error = np.mean(
    np.abs(eta_sol - eta_theory_sol) / eta_theory_sol) * 100
print(f'Mean relative error: {relative_error:.2f}%')
```

```python
def exp_func(t, eta_0_fit, n_fit):
    return eta_0_fit * np.exp(n_fit * t)

popt, _ = curve_fit(exp_func, t_sol, eta_sol, p0=[eta_0, 3.5])
eta_0_fit, n_fit = popt

print(f'Theoretical growth rate: {n_theory:.4f}')
print(f'Numerical growth rate:   {n_fit:.4f}')
print(f'Growth rate error: {abs(n_fit - n_theory) / n_theory * 100:.2f}%')
```

```python
eta_phase = sol.y[0]
eta_dot_phase = sol.y[1]
delta_rho = -(rho_2 - rho_1) * eta_phase / lam
w_phase = eta_dot_phase

plt.figure(figsize=(6, 6))
plt.plot(delta_rho, w_phase, 'b-', linewidth=2)
plt.scatter(delta_rho[0], w_phase[0], color='green', 
            zorder=5, s=100, label='Start (t=0)')
plt.scatter(delta_rho[-1], w_phase[-1], color='red', 
            zorder=5, s=100, label=f'End (t={t_end})')
plt.xlabel('Density perturbation δρ')
plt.ylabel('Vertical velocity w')
plt.title('Phase Plane: δρ vs w at Interface')
plt.legend()
plt.grid(True)
plt.show()
```

```python
eta_range = np.linspace(-0.2, 0.2, 20)
eta_dot_range = np.linspace(-1, 1, 20)
ETA, ETA_DOT = np.meshgrid(eta_range, eta_dot_range)
DETA = ETA_DOT
DETA_DOT = g * k * A * ETA
magnitude = np.sqrt(DETA**2 + DETA_DOT**2)
magnitude[magnitude == 0] = 1

plt.figure(figsize=(8, 6))
plt.quiver(ETA, ETA_DOT, DETA/magnitude, DETA_DOT/magnitude, 
           magnitude, cmap='RdBu', alpha=0.8)
plt.plot(eta_sol, np.gradient(eta_sol, t_sol), 
         'b-', linewidth=2, label='Numerical trajectory')
plt.scatter(0, 0, color='black', s=100, zorder=5, label='Fixed point (saddle)')
manifold = np.linspace(-0.2, 0.2, 100)
plt.plot(manifold, np.sqrt(g*k*A) * manifold, 
         'r--', linewidth=1.5, label='Unstable manifold (λ₁ = +3.51)')
plt.plot(manifold, -np.sqrt(g*k*A) * manifold, 
         'g', linewidth=1.5, label='Stable manifold (λ₂ = -3.51)')
plt.xlabel('Interface amplitude η')
plt.ylabel('Growth rate η̇')
plt.title(f'Phase Portrait, RTI(A = {A:.2f})')
plt.legend(loc='upper left')
plt.colorbar(label='Vector magnitude')
plt.grid(True)
plt.axhline(y=0, color='k', linewidth=0.5)
plt.axvline(x=0, color='k', linewidth=0.5)
plt.show()
```
